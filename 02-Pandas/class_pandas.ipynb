```python
e=pd.read_csv('https://raw.githubusercontent.com/whoknowayush/DataAnalysis-Dataset/refs/heads/main/Datasets-Practice/employee_data.csv')
e
```

```python
print(e.shape,'\n')
print(e.info())
```

```python
# to fill the age col by takimg mean of age

mean=e['age'].mean()
e['age']=e['age'].fillna(mean).astype(int)
e
```

```python
# to assing gender for row no 8 ita possible in small data we cant assume some ones gender or cant take mean

e.loc[8,'gender']='Female'
e
```

```python
e['gender']=e['gender'].str.capitalize()
e
```

```python
# Samuel L. Jackson have no sal to fill it we take avg sal of I.T dept

mean_IT_sal=e.groupby('department').get_group('IT')['salary'].mean()
e['salary']=e['salary'].fillna(mean_IT_sal)
e
```

```python
e['join_date']=e['join_date'].fillna("")

mean_exp=e['exp_years'].mean()

e['exp_years']=e['exp_years'].fillna(mean_exp).astype(int)

e['emp_status']=e['emp_status'].fillna(method='backfill')
e
```

```python
# Rename columns to Upper Case
e.columns = e.columns.str.upper()
e
```

```python
# Filter active employees with experience ≥ 5 years and salary > 50k
active_employees=e[(e['EXP_YEARS']>=5) & (e['SALARY']>50000)]
active_employees
```

```python
#  Get employees who joined after 2018 and are in ‘hr’ or ‘finance’

e['JOIN_DATE'] = pd.to_datetime(e['JOIN_DATE'])  # ensure datetime format

e[(e['JOIN_DATE']>='2019-01-01') & (e['DEPARTMENT']== 'HR') | (e['DEPARTMENT']=='Sales') ]
```

```python
# 1. years_with_company
e['YEARS_WITH_COMPANY'] = 2025 - e['JOIN_DATE'].dt.year
e
```

```python
# 2. is_senior flag (>=10 years experience)
e['IS_SENIOR'] = e['EXP_YEARS'] >= 10
e
```

```python
# 3. Extract first_name and last_name
e['FIRST_NAME'] = e['FULL_NAME'].str.split().str[0]
e['LAST_NAME'] = e['FULL_NAME'].str.split().str[-1]
e
```

```python
# 4. Experience-to-age ratio
e['EXP_AGE_RATIO'] = e['EXP_YEARS'] / e['AGE']
e
```

```python
# 5. Name length
e['NAME_LENGTH'] = e['FULL_NAME'].str.len()
e
```

```python
# 6. Combined feature: dept_seniority
e['DEPT_SENIORITY'] = e['DEPARTMENT'] + '_' + e['IS_SENIOR'].astype(str)
e
```

```python
# 7. salary_relative = salary / dept_avg_salary
dept_avg_salary = e.groupby('DEPARTMENT')['SALARY'].transform('mean')
e['SALARY_RELATIVE'] = e['SALARY'] / dept_avg_salary
e
```

```python
# Set EMP_ID as Index

e.set_index('EMP_ID', inplace=True)
e
```

```python
# Reset Index Back to Default

e.reset_index(inplace=True)
e.head()
```
